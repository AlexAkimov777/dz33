```python
# спросил у нейронки про чтение файла и она сказала что по pep8 правильней всего именно так делить список на строки
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_csv('iris_data.csv')
spec = data['Species'].to_numpy()
k1 = np.sum(spec == 'Iris-setosa')
k2 = np.sum(spec == 'Iris-versicolor')
k3 = np.sum(spec == 'Iris-virginica')
ln = data['PetalLengthCm'].to_numpy()
l1 = np.sum(ln > 1.2)
l2 = np.sum((ln > 1.2) & (ln < 1.5))
l3 = np.sum(ln > 1.5)
plt.figure(figsize=(20, 20), dpi=100)
plt.subplot(1, 2, 1)
plt.pie(
    [k1, k2, k3],
    labels=[
        'Iris-setosa',
        'Iris-versicolor',
        'Iris-virginica'
    ]
)
plt.title('Species')
plt.subplot(1, 2, 2)
plt.pie(
    [l1, l2, l3],
    labels=[
        'PetalLengthCm > 1.2',
        '1.2 < PetalLengthCm < 1.5',
        'PetalLengthCm > 1.5'
    ]
)
plt.title('PetalLengthCm')
plt.show()
```

```python
# функцию для мнк я писал на прошлой неделе
# нейронка помогла с выводом данных на график 
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_csv('iris_data.csv')
lnsepal = data['SepalLengthCm'].to_numpy()
wdsepal = data['SepalWidthCm'].to_numpy()
lnpetal = data['PetalLengthCm'].to_numpy()
wdpetal = data['PetalWidthCm'].to_numpy()


def mnk(x, y):
    x = np.array(x, dtype=float)
    y = np.array(y, dtype=float)
    n = len(x)
    sredx = np.sum(x) / n
    sredy = np.sum(y) / n
    sredx2 = np.sum(x ** 2) / n
    sredxy = np.sum(x * y) / n
    k = (sredxy - sredx * sredy) / (sredx2 - sredx ** 2)
    b = sredy - k * sredx
    return k, b


plt.figure(figsize=(15, 15), dpi=100)
plt.subplot(2, 3, 1)
plt.scatter(lnsepal, wdsepal)
plt.title('SepalLengthCm от SepalWidthCm')
plt.subplot(2, 3, 2)
plt.scatter(lnsepal, lnpetal)
plt.title('SepalLengthCm от PetalLengthCm')
plt.subplot(2, 3, 3)
plt.scatter(lnsepal, wdpetal)
plt.title('SepalLengthCm от PetalWidthCm')
plt.subplot(2, 3, 4)
plt.scatter(wdsepal, lnpetal)
plt.title('SepalWidthCm от PetalLengthCm')
plt.subplot(2, 3, 5)
plt.scatter(wdsepal, wdpetal)
plt.title('SepalWidthCm от PetalWidthCm')
plt.subplot(2, 3, 6)
k, b = mnk(lnpetal, wdpetal)
plt.scatter(lnpetal, wdpetal)
x1 = np.array([np.min(lnpetal), np.max(lnpetal)])
y1 = k * x1 + b
plt.plot(x1, y1)
plt.title('PetalLengthCm от PetalWidthCm')
plt.text(1.5, 2.2, f'k = {k:.2f}\nb = {b:.2f}')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_csv('BTC_data.csv')
close = data['close'].to_numpy(dtype=int)
time = data['time'].to_numpy()
for i in range(len(time)):
    a = time[i][:10]
    a = a.split('-')
    a = a[::-1]
    time[i] = '-'.join(a)
plt.figure(figsize=(20, 15), dpi=100)
plt.plot(time, close)
plt.xticks(time[::100])
plt.show()
```
