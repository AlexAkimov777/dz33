```python
def perevod(a):
    if len(a) >= 4:
        k = 0
        for i in range(4):
            if a[i] == a[i].upper() and not a[i].isdigit():
                k += 1
        if k >= 3:
            print(a.upper())
        else:
            print(a)
    else:
        print(a)


a = input()
perevod(a)
```

```python
# спросил у нейронки про условие
def html(t, s):
    if t not in [
        'a', 'abbr', 'b', 'body', 'caption', 'cite',
        'code', 'div', 'form', 'h1', 'h2', 'h3', 'h4',
        'h5', 'h6', 'header', 'i', 's'
    ]:
        print('Введён неверный тег')
    else:
        print(f'<{t}>{s}</{t}>')


t = input()
s = input()
html(t, s)
```

```python
a = 'dfas'
n = len(a)
if n == 0:
    print(0)
elif n <= 2:
    print(ord(a[0]))
elif 2 < n <= 10:
    if n % 2 == 0:
        print(ord(a[0]) + ord(a[n // 2 - 1]) + ord(a[n - 1]))
    else:
        print(ord(a[0]) + ord(a[(n - 1) // 2]) + ord(a[n - 1]))
else:
    print(ord(a[n - 1]))
```
