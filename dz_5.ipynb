```python
#спросил у нейронки насчет обработки данных от точек и т.д.
from string import punctuation

with open("/usr/share/licenses/python/LICENSE", "r") as f:
    t = f.read()
t = t.lower()
for s in punctuation:
    t = t.replace(s, " ")
a = t.split()
p = []
c = dict()
for i in a:
    if i not in p:
        p.append(i)
        c[i] = a.count(i)
k = list(c.keys())
v = list(c.values())
for _ in range(10):
    mav = max(v)
    ind = v.index(mav)
    print(k[ind], mav)
    del k[ind]
    v.remove(mav)

```

```python
s1 = [1, 4, 546, 324, 245, 67, 234, 789, 234, 7]
s2 = [23, 54, 7, 45, 435, 67, 98, 23, 4958, 34, 8796, 345, 7]
m1 = set(s1)
m2 = set(s2)
print('Уникальные для 1 списка:', *list(m1))
print('Уникальные для 2 списка:', *list(m2))
print(
    'Уникальные для объединения спиков:',
    *list(m1.union(m2))
)
print(
    'Уникальные, содержащиеся в обоих списках:',
    *list(m1.intersection(m2))
)
```

```python
a = input()
s = a.split('student_')
del s[0]
s1 = []
s2 = []
for i in range(len(s)):
    s2.append(int(s[i][3:]))
    s1.append(s[i][:3])
ma = max(s2)
ind = s2.index(ma)
k = s2.count(ma)
if k == 1:
    print(s1[ind])
else:
    r = []
    for _ in range(k):
        ind = s2.index(ma)
        r.append(s1[ind])
        del s1[ind]
        del s2[ind]
    print('-'.join(r))
```

```python
#спросил у нейронке про округление
import numpy as np

s = input().split()
r = float(s[0])
a = float(s[1])
l = 2 * np.pi * r
if int(str(l).split('.')[1][2]) != 5:
    l = str(round(l, 2))
else:
    l = str(round(l + 0.001, 2))
s1 = np.pi * r ** 2
s2 = a ** 2
pr = s1 / s2 * 100
if int(str(pr).split('.')[1][2]) != 5:
    pr = str(round(pr, 2))
else:
    pr = str(round(pr + 0.001, 2))
print(
    'Длина окружности равно', l + '.',
    'Площадь круга составляет', pr + '%',
    'от площади квадрата.'
)
```
